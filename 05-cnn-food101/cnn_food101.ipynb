## **1.Descripción**  



### 1.1 **Fundamentos Teóricos**

Las redes neuronales convolucionales (CNN, por sus siglas en inglés) son una clase de modelos de aprendizaje profundo diseñados específicamente para procesar datos con una estructura en forma de cuadrícula, como las imágenes. Su arquitectura se inspira en el funcionamiento del córtex visual de los animales, permitiendo la extracción automática y jerárquica de características relevantes.

Las CNN están compuestas principalmente por varios tipos de capas:

- **Capas convolucionales:** Aplican filtros o kernels que recorren la imagen para detectar patrones locales, como bordes, texturas o formas. Cada filtro aprende a identificar una característica específica durante el entrenamiento.
- **Capas de activación:** Introducen no linealidades en el modelo, siendo la función ReLU (Rectified Linear Unit) la más utilizada, lo que permite a la red aprender representaciones complejas.
- **Capas de pooling:** Reducen la dimensionalidad de las representaciones intermedias, conservando la información más relevante y disminuyendo el costo computacional. El pooling más común es el max pooling.
- **Capas completamente conectadas:** Al final de la red, estas capas integran la información extraída para realizar la clasificación o regresión.

El entrenamiento de una CNN se realiza mediante algoritmos de optimización como el descenso por gradiente, ajustando los pesos de los filtros para minimizar una función de pérdida. El uso de técnicas como la normalización, regularización y aumento de datos contribuye a mejorar la generalización del modelo.

Las CNN han revolucionado el campo de la visión por computadora, logrando resultados sobresalientes en tareas como clasificación de imágenes, detección de objetos y segmentación, entre otras.

#### Carga y preparación de los datos para entrenamiento y validación

En este bloque de código se implementa una función para cargar y preparar el dataset Food-101, dividiéndolo en conjuntos de entrenamiento y validación, y normalizando las imágenes para su uso en modelos de redes neuronales convolucionales.

- Se utiliza TensorFlow para cargar las imágenes desde el directorio especificado, dividiendo automáticamente el dataset en un 80% para entrenamiento y un 20% para validación mediante el parámetro `validation_split`.
- Las imágenes se redimensionan a un tamaño estándar (por ejemplo, 100x100 píxeles) y se agrupan en lotes (`batch_size`) para optimizar el entrenamiento.
- Se aplica una capa de normalización para escalar los valores de los píxeles al rango [0, 1], lo que mejora la estabilidad y eficiencia del aprendizaje.
- El conjunto de entrenamiento se mezcla (shuffle) para favorecer la generalización del modelo.
- Finalmente, se imprime un ejemplo de las clases presentes y el número total de categorías en el dataset.

**Resultado esperado:**  
Se obtienen dos conjuntos de datos listos para ser utilizados en el entrenamiento y validación de una red neuronal, junto con la lista de clases disponibles. Esto sienta las bases para construir y entrenar modelos de clasificación de imágenes de manera eficiente y reproducible.



````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

def load_food101_data(path, image_size=(125, 125), batch_size=32):
    # Carga con división train/validation
    train_ds_raw = tf.keras.utils.image_dataset_from_directory(
        path,
        validation_split=0.2,
        subset="training",
        seed=123,
        image_size=image_size,
        batch_size=batch_size
    )

    val_ds = tf.keras.utils.image_dataset_from_directory(
        path,
        validation_split=0.2,
        subset="validation",
        seed=123,
        image_size=image_size,
        batch_size=batch_size
    )

    # Normalizar imágenes (de 0-255 a 0-1)
    normalization_layer = tf.keras.layers.Rescaling(1./255)
    train_ds = train_ds_raw.map(lambda x, y: (normalization_layer(x), y))
    val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

    # Shuffle en entrenamiento para mejor generalización
    train_ds = train_ds.shuffle(1000)

    return train_ds, val_ds, train_ds_raw.class_names

# Ruta a tu dataset (ajusta esta ruta)
path = "food-101/images"

# Carga datos
train_ds, val_ds, class_names = load_food101_data(path, image_size=(125, 125), batch_size=64)

# Verifica las clases
print("Clases (ejemplo):", class_names[:10])
print("Número total clases:", len(class_names))
````

# Modelo CNN desde Cero (Sin Transfer Learning)

¿Qué es?
Este modelo es una **red neuronal convolucional (CNN)** diseñada y entrenada desde cero, sin utilizar pesos preentrenados ni conocimiento previo. Es decir, **no aplica Transfer Learning**. Todo el aprendizaje se realiza directamente desde los datos del proyecto.

---

Características Clave del Modelo

- **Arquitectura personalizada**: Compuesta por 3 bloques convolucionales con capas de normalización, activación ReLU y técnicas de regularización como Dropout y L2.
- **Entrenamiento desde cero**: El modelo comienza con pesos aleatorios y aprende directamente del dataset.
- **Clasificación multiclase**: Diseñado para clasificar imágenes en **101 clases diferentes**.
- **Callbacks inteligentes**:
  - `EarlyStopping`: Detiene el entrenamiento si no hay mejora, evitando sobreajuste.
  - `ModelCheckpoint`: Guarda automáticamente el mejor modelo.

---

¿Por qué es importante?

- **Mayor control**: Permite diseñar una arquitectura adaptada 100% al problema específico.
- **Sin dependencia externa**: No se basa en modelos preentrenados, útil cuando el dominio de los datos es muy diferente a datasets comunes como ImageNet.
- **Punto de comparación**: Sirve como línea base para comparar con modelos que sí usan Transfer Learning o Fine-Tuning.

---

¿Cuándo usar este enfoque?

- Cuando se dispone de **una gran cantidad de datos etiquetados**.
- Cuando el dominio de las imágenes es **muy específico o diferente** a los modelos preentrenados.
- Para **experimentar y aprender** sobre el diseño de redes neuronales desde cero.


````python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Modelo CNN personalizado
def build_custom_model(input_shape=(125, 125, 3), num_classes=101):
    model = models.Sequential([
        # Bloque 1
        layers.Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=input_shape),
        layers.BatchNormalization(),
        layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        # Bloque 2
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.3),
        
        # Bloque 3
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.4),
        
        # Clasificador
        layers.Flatten(),
        layers.Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    
    return model

# Compilación del modelo
model = build_custom_model()
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Callbacks
callbacks = [
    EarlyStopping(patience=5, restore_best_weights=True),
    ModelCheckpoint('custom_mejorado.h5', save_best_only=True)
]
````

# Modelo con Transfer Learning (MobileNetV2)

¿Qué es?
Este modelo utiliza **Transfer Learning**, una técnica que aprovecha un modelo previamente entrenado (en este caso, **MobileNetV2** entrenado con ImageNet) para resolver una nueva tarea de clasificación. Esto permite **acelerar el entrenamiento** y mejorar el rendimiento, especialmente cuando se dispone de pocos datos.

---

Características Clave del Modelo

- **Modelo base preentrenado**: Se utiliza MobileNetV2 sin su capa final (`include_top=False`) y con pesos entrenados en ImageNet.
- **Congelamiento inicial**: Las capas del modelo base están congeladas al inicio para preservar el conocimiento aprendido.
- **Capas superiores personalizadas**:
  - `GlobalAveragePooling2D` para reducir la dimensionalidad.
  - Capas densas con regularización y Dropout para evitar sobreajuste.
  - Capa final `softmax` adaptada al número de clases del nuevo dataset.

---

¿Por qué es importante?

- **Aprovecha conocimiento previo**: El modelo ya ha aprendido a detectar patrones generales en imágenes (bordes, texturas, formas).
- **Entrenamiento más rápido y eficiente**: Requiere menos datos y menos tiempo para alcanzar buenos resultados.
- **Mejor rendimiento inicial**: Suele comenzar con una precisión más alta que un modelo entrenado desde cero.

---

¿Cuándo usar este enfoque?

- Cuando se dispone de **pocos datos etiquetados**.
- Cuando se quiere **reducir el tiempo de entrenamiento**.
- Cuando el dominio de las imágenes es **relativamente similar** al de los datos con los que se entrenó el modelo base (por ejemplo, objetos naturales, escenas, etc.).

---

Callbacks utilizados

- `EarlyStopping`: Detiene el entrenamiento si no hay mejora en la validación.
- `ModelCheckpoint`: Guarda el mejor modelo basado en la precisión de validación.
- `CSVLogger`: Registra el historial del entrenamiento para análisis posterior.


````python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, CSVLogger

# Modelo base MobileNetV2 preentrenado
base_model = MobileNetV2(
    input_shape=(125, 125, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False  # Congelado para entrenamiento inicial

# Construcción del modelo
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.BatchNormalization(),

    layers.Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    layers.BatchNormalization(),
    layers.Dropout(0.5),

    layers.Dense(len(class_names), activation='softmax')
])

# Compilar modelo
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Callbacks
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
checkpoint = ModelCheckpoint('modelo_base.h5', monitor='val_accuracy', save_best_only=True)
csv_logger = CSVLogger('entrenamiento_historial_Final.csv')


````


#### Entrenamiento del modelo CNN

En este bloque de código se entrena el modelo de red neuronal convolucional previamente definido utilizando los conjuntos de datos de entrenamiento y validación.

- Se utiliza el método `fit` para ajustar el modelo a los datos de entrenamiento (`train_ds`) durante un número determinado de épocas (en este caso, 8).

- El parámetro `verbose=1` permite visualizar el progreso del entrenamiento en la consola.

**Resultado esperado:**  
El modelo se entrena durante varias épocas, mostrando la evolución de la pérdida y la precisión tanto en entrenamiento como en validación. Al finalizar, se dispone del mejor modelo guardado y se evitan entrenamientos innecesarios si el modelo deja de mejorar.



````python
# Entrenamiento
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=[early_stop, checkpoint, csv_logger]
)
````


## Explicación: Predicción de Imágenes 

Este bloque de código permite realizar la predicción automática de la clase de todas las imágenes contenidas en una carpeta, utilizando un modelo previamente entrenado y guardado. Es ideal para analizar múltiples imágenes de prueba de manera eficiente y visualizar los resultados de forma clara.

### ¿Cómo funciona la función `predict_images_in_folder`?

1. **Carga del modelo:**
   - Se carga el modelo de red neuronal previamente entrenado desde el archivo especificado (`model_path`).

2. **Búsqueda de imágenes:**
   - Se listan todos los archivos de imagen en la carpeta indicada (`folder_path`). Se consideran archivos con extensiones `.png`, `.jpg` y `.jpeg`.

3. **Procesamiento y predicción:**
   - Para cada imagen encontrada:
     - Se carga y redimensiona al tamaño requerido por el modelo (`img_size`).
     - Se normalizan los valores de los píxeles al rango [0, 1].
     - Se realiza la predicción con el modelo cargado.
     - Se obtiene la clase predicha y el nivel de confianza (porcentaje de probabilidad máxima).

4. **Visualización y reporte:**
   - Se muestra cada imagen junto con el nombre del archivo, la clase predicha y el porcentaje de confianza en el título.
   - Se imprime en consola un resumen con el nombre del archivo, la clase predicha y la confianza.

### Ventajas de este enfoque

- Permite analizar rápidamente grandes lotes de imágenes sin intervención manual.
- Facilita la validación del modelo sobre datos nuevos o no vistos durante el entrenamiento.
- La visualización directa ayuda a identificar aciertos y errores de manera intuitiva.

### Ejemplo de uso

Solo debes indicar la ruta del modelo guardado, la carpeta con las imágenes a analizar y la lista de nombres de clases:

```python
predict_images_in_folder(
    model_path="modelo_finetuned_final.h5",
    folder_path="ImagenesPrueba", 
    class_names=class_names,
    img_size=(125, 125)
)
```

### Interpretación de resultados

- Cada imagen se muestra con su predicción y el nivel de confianza.
- En la consola, verás un resumen por archivo.
- Si la confianza es alta, el modelo está seguro de su predicción; si es baja, puede haber ambigüedad o similitud con otras clases.

Esta función es especialmente útil para evaluar el desempeño del modelo en escenarios reales y para preparar reportes visuales de resultados.




````python
import os
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt

def predict_images_with_multiple_models(models_info, folder_path, class_names, img_size=(125, 125)):
    """
    Carga varios modelos y predice la clase de todas las imágenes en una carpeta, mostrando la predicción de cada modelo.
    models_info: lista de tuplas (nombre, ruta_modelo)
    """
    # Cargar todos los modelos
    modelos = [(nombre, load_model(ruta)) for nombre, ruta in models_info]
    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    for img_file in image_files:
        img_path = os.path.join(folder_path, img_file)
        img = image.load_img(img_path, target_size=img_size)
        img_array = image.img_to_array(img).astype("float32") / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        pred_texts = []
        for nombre, modelo in modelos:
            preds = modelo.predict(img_array)
            pred_class = np.argmax(preds, axis=1)[0]
            class_name = class_names[pred_class]
            confidence = 100 * np.max(preds)
            pred_texts.append(f"{nombre}: {class_name} ({confidence:.2f}%)")

        # Mostrar imagen y predicciones
        plt.figure(figsize=(6, 4))
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"{img_file}\n" + "\n".join(pred_texts), fontsize=10)
        plt.tight_layout()
        plt.show()

        # Imprimir en consola también
        print(f"{img_file}:")
        for txt in pred_texts:
            print("  " + txt)
        print("-" * 40)

# Ejemplo de uso:
models_info = [
    ("Modelo Base", "modelo_mejorado.h5"),
    ("Modelo TransferLearning", "modelo_base.h5"),
    ("Modelo Fine-Tuned", "modelo_finetuned_final.h5")
]
predict_images_with_multiple_models(
    models_info=models_info,
    folder_path="ImagenesPrueba",
    class_names=class_names,
    img_size=(125, 125)
)
````




### Explicación del Resultado

Al ejecutar la función de predicción sobre una carpeta de imágenes, se obtiene para cada archivo la clase predicha y el nivel de confianza del modelo. Por ejemplo, si una imagen es clasificada como **"chocolate_cake"** con una confianza del **46.77%**, esto nos permite analizar varios aspectos:

- **Interpretación:**  
  El modelo selecciona la clase con mayor probabilidad, pero un valor de confianza bajo (por ejemplo, menor al 50%) indica que el modelo no está completamente seguro. Esto puede deberse a similitud visual entre clases, calidad de la imagen, o características ambiguas.

- **Utilidad:**  
  Estos resultados permiten evaluar la capacidad del modelo para generalizar sobre imágenes nuevas y detectar casos donde el modelo duda o puede confundirse. Es útil para identificar imágenes problemáticas, clases difíciles o posibles errores de predicción.

- **Acciones recomendadas:**  
  - Revisar imágenes con baja confianza para ver si realmente son difíciles o si hay errores en el etiquetado.
  - Considerar aumentar la cantidad de datos de entrenamiento para las clases con más errores o baja confianza.
  - Aplicar técnicas de aumento de datos (data augmentation) o probar arquitecturas más robustas si el problema persiste.
  - Analizar si las imágenes problemáticas tienen condiciones de iluminación, ángulos o calidad diferentes a las del entrenamiento.

- **Conclusión:**  
  La función de predicción por carpeta no solo permite clasificar imágenes en lote, sino que también es una herramienta valiosa para auditar el desempeño real del modelo, identificar sus límites y planificar mejoras futuras.





## Explicación: Predicción por Ensemble de Modelos 

**¿Cómo funciona?**

1. **Carga de imágenes:**  
   - Se buscan todas las imágenes en la carpeta indicada (`folder_path`) con extensiones `.png`, `.jpg` o `.jpeg`.

2. **Predicción con múltiples modelos:**  
   - Para cada imagen, se carga y preprocesa (redimensiona y normaliza).
   - Se recorre la lista de rutas de modelos (`models_paths`), cargando cada modelo y obteniendo su predicción para la imagen.
   - Todas las predicciones se almacenan en una lista.

3. **Promedio de predicciones (Ensemble):**  
   - Se calcula el promedio de las predicciones de todos los modelos para cada clase (votación suave o *soft voting*).
   - Se selecciona la clase con mayor probabilidad promedio como la predicción final.

4. **Reporte de resultados:**  
   - Se imprime el nombre del archivo, la clase predicha y el nivel de confianza promedio (porcentaje de la probabilidad máxima).



**¿Para qué sirve?**

- El *ensemble* combina la "opinión" de varios modelos, lo que suele mejorar la robustez y precisión de las predicciones en comparación con usar un solo modelo.
- Permite reducir el impacto de errores o sesgos de un modelo individual, aprovechando la diversidad de arquitecturas o entrenamientos.



**¿Qué mejora respecto a una sola red?**

- Suele aumentar la precisión y la confianza en la predicción, especialmente si los modelos son diferentes o han sido entrenados con distintos datos o hiperparámetros.
- Hace el sistema más resistente a errores puntuales de un modelo, ya que la decisión final es el consenso de varios.



**Resumen:**  
Este enfoque es ideal cuando se dispone de varios modelos entrenados y se busca obtener la mejor predicción posible para un conjunto de imágenes, aprovechando la fortaleza colectiva de todos los modelos.



````python
import os
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image

def ensemble_predict_folder(models_paths, folder_path, class_names, img_size=(125, 125)):
    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    for img_file in image_files:
        img_path = os.path.join(folder_path, img_file)
        preds_list = []
        img = image.load_img(img_path, target_size=img_size)
        img_array = image.img_to_array(img).astype("float32") / 255.0
        img_array = np.expand_dims(img_array, axis=0)
        for path in models_paths:
            model = load_model(path)
            preds = model.predict(img_array)
            preds_list.append(preds[0])
        avg_preds = np.mean(preds_list, axis=0)
        pred_class = np.argmax(avg_preds)
        class_name = class_names[pred_class]
        confidence = 100 * np.max(avg_preds)
        print(f'{img_file}: {class_name} ({confidence:.2f}%)')

# Ejemplo de uso:
ensemble_predict_folder(
    models_paths=["modelo_mejorado.h5","modelo_base.h5", "modelo_finetuned_final.h5"],
    folder_path="ImagenesPrueba",
    class_names=class_names
)
````

- **Utilidad:**
